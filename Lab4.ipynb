```python
def bfs(startingNode, destinationNode, G):
    reached = []  # to store reached (generated) nodes
    expand_sequence = []  # to store the order of nodes expanded
    frontier = []

    snode = {'City': startingNode, 'Path cost': 0, 'Path': [startingNode]}
    frontier.append(snode)    # insert the starting node
    reached.append(startingNode)

    while len(frontier) > 0:
        print(frontier)
        unode = frontier.pop(0)  # selecting the node to expand
        u = unode['City']
        expand_sequence.append(u)
        print(u + " expanding..")

        for v in G[u].keys():
            if v not in reached:  # process if v is not reached (generated) yet
                print(v + " generated.")
                cost = unode['Path cost'] + G[u][v]
                path = unode['Path'] + [v]
                vnode = {'City': v, 'Path cost': cost, 'Path': path}
                frontier.append(vnode)  # store the new child node in the frontier
                reached.append(v)
                # goal test
                if v == destinationNode:
                    print(expand_sequence)
                    return vnode  # goal found
    print('Failed')
```

```python
bfs('Arad', 'Bucharest', romanian_map)
```

```python
def dfs(startingNode, destinationNode, G):
    reached = []          # visited nodes
    expand_sequence = []  # nodes expansion order
    frontier = []

    snode = {'City': startingNode, 'Path cost': 0, 'Path': [startingNode]}
    frontier.append(snode)

    while len(frontier) > 0:
        unode = frontier.pop()  # LIFO for DFS (stack)
        u = unode['City']
        expand_sequence.append(u)

        if u not in reached:
            reached.append(u)

            for v in G[u].keys():
                if v not in reached:
                    cost = unode['Path cost'] + G[u][v]
                    path = unode['Path'] + [v]
                    vnode = {'City': v, 'Path cost': cost, 'Path': path}
                    frontier.append(vnode)

                    if v == destinationNode:
                        print("DFS Expansion Sequence:", expand_sequence)
                        return vnode
    print('Failed')
    return None
```

```python
bfs_result = bfs('Arad', 'Bucharest', romanian_map)
print("BFS Result:", bfs_result)
```

```python
dfs_result = dfs('Arad', 'Bucharest', romanian_map)
print("DFS Result:", dfs_result)
```
